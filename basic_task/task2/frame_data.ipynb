## Imports


```python
import pandas as pd
import numpy as np
import math
```

## Dictionary to store

```python
data = {'column1':[1,2,3,4,5],
       'column2': [3,2,4,1,5]
     
       } 
```

## Creating DataFrame for csv file


```python
df = pd.DataFrame(data=data)
```

```python
df.to_csv("data.csv")
```

## Reading file from csv

```python
df2 = pd.read_csv("data.csv")
```

## Now we have to covert column to list for applying algorithms

```python
column1_list = df2['column1'].tolist()
column2_list = df2['column2'].tolist()
```

## Applying bubble sort algorithm

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
```

## Applying linear search

```python
 def linear_search(arr1, arr2, target):
    results = {
        'column1': [],
        'column2': []
    }

    # Search in the first column
    for i in range(len(arr1)):
        if arr1[i] == target:
            results['column1'].append(i)  

    # Search in the second column
    for i in range(len(arr2)):
        if arr2[i] == target:
            results['column2'].append(i) 

    return results  
```

## applying sorting algo on column1 and column2

```python
sorted_column1 = bubble_sort(column1_list.copy())
sorted_column2 = bubble_sort(column2_list.copy())
```

```python
print("Sorted 'column1' (Bubble Sort):", sorted_column1)
print("Sorted 'column2' (Bubble Sort):", sorted_column2)
```

## target value dengay ju humne search karni hai


```python
target_value = 4
search_results = linear_search(column1_list, column2_list, target_value)
```

```python

print(f"\nWhere 'column1' equals {target_value} (Linear Search):", search_results['column1'])
print(f"\nWhere 'column2' equals {target_value} (Linear Search):", search_results['column2'])
```

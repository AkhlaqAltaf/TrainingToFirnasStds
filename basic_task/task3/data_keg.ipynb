## Applying merge sort algorithm

```python

# Merge Sort Function
def merge_sort(df, column):
    if len(df) > 1:
        mid = len(df) // 2  # Finding the mid of the array
        left_half = df.iloc[:mid]  
        right_half = df.iloc[mid:]

       
        merge_sort(left_half, column)
        merge_sort(right_half, column)

        i = j = k = 0

        # Merging the sorted halves
        while i < len(left_half) and j < len(right_half):
            if left_half[column].iloc[i] <= right_half[column].iloc[j]:
                df.iloc[k] = left_half.iloc[i]  
                i += 1
            else:
                df.iloc[k] = right_half.iloc[j]
                j += 1
            k += 1

        # Checking if any element was  at left
        while i < len(left_half):
            df.iloc[k] = left_half.iloc[i]
            i += 1
            k += 1
        #check if any element was at right
        while j < len(right_half):
            df.iloc[k] = right_half.iloc[j]
            j += 1
            k += 1


merge_sort(df, 'Age')


sorted_df = df[['Age', 'User Behavior Class']]

# Display the sorted columns 
print(sorted_df)
```
